```python
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv("/kaggle/input/wine-quality-dataset/WineQT.csv")
df.head()
```

```python
# Displaying a scatter plot to compare the difference between fixed acidity and volatile acidity
plt.figure(figsize=(8, 6))
sns.scatterplot(x='fixed acidity', y='volatile acidity', data=df, color='green')
plt.title('Fixed Acidity vs Volatile Acidity Scatter Plot')
plt.xlabel('Fixed Acidity Levels')
plt.ylabel('Volatile Acidity Levels')
plt.show()
```

```python
# Visualizing the relationship between total sulfur dioxide and density
plt.figure(figsize=(8, 6))
sns.scatterplot(x='total sulfur dioxide', y='density', data=df, color='orange')
plt.title('Total Sulfur Dioxide vs Density Scatter Plot')
plt.xlabel('Total Sulfur Dioxide Levels')
plt.ylabel('Density Levels')
plt.show()
```

```python
# Visualizing the relationship between density and fixed acidity
import seaborn as sns
import matplotlib.pyplot as plt

# Scatter plot for density vs acidity
plt.figure(figsize=(8, 6))
sns.scatterplot(x='density', y='fixed acidity', data=df, color='blue')
plt.title('Scatter Plot of Density vs Fixed Acidity')
plt.xlabel('Density')
plt.ylabel('Fixed Acidity')
plt.show()
```

```python
# Excluding the last three columns from the dataset and printing the modified DataFrame
import pandas as pd
df = pd.read_csv("/kaggle/input/wine-quality-dataset/WineQT.csv")
df = df.iloc[:, :-3]
print(df)
```

```python
print(df.columns)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Visualizing the distribution of fixed acidity using KDE plot
plt.figure(figsize=(8, 6))
sns.kdeplot(data=df['fixed acidity'], color='pink', shade=True)
plt.title('Distribution of Fixed Acidity')
plt.xlabel('Fixed Acidity Levels')
plt.ylabel('Density')
plt.show()
```

```python
# Grouping by pH and calculating mean and sum for density and sulphates
import seaborn as sns
import matplotlib.pyplot as plt

grouped_df = df.groupby('pH').agg({'density': 'mean', 'sulphates': 'sum'}).reset_index()
print(grouped_df.head())

# Visualizing the relationship between pH and density using a scatter plot
plt.figure(figsize=(8, 6))
sns.scatterplot(x='pH', y='density', data=grouped_df, color='red')
plt.title('Mean Density vs pH')
plt.xlabel('pH')
plt.ylabel('Mean Density')
plt.show()
```
